```python
import pandas as pd
import numpy as np

penguins = pd.read_csv('../data/raw/penguins.csv')
penguins.head()
```

```python
col1 = 'bill_length_mm'
col2 = 'body_mass_g'

df = penguins.copy()

n_rows = len(df)
n_missing = int(n_rows * 0.10)
print(f"Будет внесено пропусков (NaN) в каждую колонку: {n_missing}")

np.random.seed(42)
idx_col1 = np.random.choice(n_rows, size=n_missing, replace=False)
idx_col2 = np.random.choice(n_rows, size=n_missing, replace=False)

df.loc[idx_col1, col1] = np.nan
df.loc[idx_col2, col2] = np.nan
```

```python
penguins.isnull().sum()
```

```python
df.isnull().sum()
```

```python
df.to_csv('../data/change/change_penguins.csv', index=False)
```
