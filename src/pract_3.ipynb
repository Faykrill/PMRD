```python
import featuretools as ft
```

```python
import pandas as pd


product = pd.read_csv('..//data//product.csv')
customer = pd.read_csv('..//data//customer.csv')
order = pd.read_csv('..//data//order.csv')
```

```python
es = ft.EntitySet(id="electronics_store")

es.add_dataframe(
    dataframe_name="customer",
    dataframe=customer,
    index="customer_id"
)

es.add_dataframe(
    dataframe_name="product",
    dataframe=product,
    index="product_id"
)

es.add_dataframe(
    dataframe_name="order",
    dataframe=order,
    index="order_id",
    time_index="order_date" 
)
```

```python
es.add_relationship(
    parent_dataframe_name="customer",
    parent_column_name="customer_id",
    child_dataframe_name="order",
    child_column_name="customer_id"
)

es.add_relationship(
    parent_dataframe_name="product",
    parent_column_name="product_id",
    child_dataframe_name="order",
    child_column_name="product_id"
)
```

```python
# Запуск Deep Feature Synthesis (DFS)
feature_matrix, feature_defs = ft.dfs(
    entityset=es,
    target_dataframe_name="customer",
    max_depth=2,
    
    # Агрегирующие примитивы (как сворачивать данные из дочерних таблиц)
    agg_primitives=["sum", "mean", "count", "max", "min", "std", "num_unique"],
    
    # Трансформирующие примитивы (как преобразовывать существующие данные, например, даты)
    trans_primitives=["year", "month", "day", "time_since_previous", "percentile"]
)

print(f"✅ Сгенерировано признаков: {feature_matrix.shape[1]}")
print(f"✅ Количество клиентов: {feature_matrix.shape[0]}\n")
```
